```python
import os
import polars as pl
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
def read_parquet_transaction(train_path: str):
    # Lấy tất cả các file parquet trong thư mục
    files = [os.path.join(train_path, f) for f in os.listdir(train_path) if f.endswith('.parquet')]
    
    # Phân loại các file theo loại tên
    purchase_chunk_files = [file for file in files if 'purchase_history_daily_chunk' in file]
        
    # Đọc các file riêng biệt thành DataFrame
    purchase_chunk_df = pl.concat([pl.read_parquet(file) for file in purchase_chunk_files]) if purchase_chunk_files else None
        
    # Trả về một dictionary chứa các DataFrame
    return purchase_chunk_df
```

```python
def read_parquet_item(train_path: str):
    # Lấy tất cả các file parquet trong thư mục
    files = [os.path.join(train_path, f) for f in os.listdir(train_path) if f.endswith('.parquet')]
    
    # Phân loại các file theo loại tên
    item_chunk_files = [file for file in files if 'item_chunk' in file]
        
    # Đọc các file riêng biệt thành DataFrame
    item_chunk_df = pl.concat([pl.read_parquet(file) for file in item_chunk_files]) if item_chunk_files else None
        
    # Trả về một dictionary chứa các DataFrame
    return item_chunk_df
```

```python
from datetime import datetime, timedelta

# Test window bạn muốn
test_start = datetime(2025, 1, 1)
test_end_excl = datetime(2025, 1, 31)  # end-exclusive để cover toàn bộ ngày 30/01

# Train window: 90 ngày trước test_start
train_end = test_start
train_start = train_end - timedelta(days=90)

print("Train:", train_start, "→", train_end)
print("Test :", test_start, "→", test_end_excl)
```

```python
train_users = first_tx_df.filter(
    (pl.col("first_tx_date") >= train_start) &
    (pl.col("first_tx_date") <  train_end)
)

test_users = first_tx_df.filter(
    (pl.col("first_tx_date") >= test_start) &
    (pl.col("first_tx_date") <  test_end_excl)
)

print("Train users:", train_users.height)
print("Test users :", test_users.height)
```

```python
train_trans_30d = (
    trans_enriched
    .join(train_users, on="customer_id", how="inner")
    .filter(
        (pl.col("created_date") >= pl.col("first_tx_date")) &
        (pl.col("created_date") <  pl.col("first_tx_date") + pl.duration(days=30))
    )
)
```

```python
K_CAT = 5

top_categories_df = (
    train_trans_30d
    .group_by("category_l1")
    .agg(pl.col("customer_id").n_unique().alias("num_users"))
    .sort("num_users", descending=True)
    .head(K_CAT)
)

top_categories = top_categories_df["category_l1"].to_list()
print("Top categories:", top_categories)
```

```python
train_first_item = (
    trans_enriched
    .join(train_users, on="customer_id", how="inner")
    .filter(pl.col("created_date") == pl.col("first_tx_date"))
    .filter(pl.col("category_l1").is_in(top_categories))
)
```

```python
K_HIT = 10        # Hit@10
K_CAT = 3         # vẫn giữ 3 category
HYBRID_ALPHA = 0.6

# Tổng = 10
# Cat #1: 5 item
# Cat #2: 3 item
# Cat #3: 2 item
alloc = [5, 3, 2]

cat_to_items = {}

for cat, n_take in zip(top_categories, alloc):
    items_cat = (
        train_first_item
        .filter(pl.col("category_l1") == cat)
        .group_by("item_id")
        .agg(pl.col("customer_id").n_unique().alias("num_users"))
        .sort("num_users", descending=True)
        .head(n_take)
        .select("item_id")
    )["item_id"].to_list()

    cat_to_items[cat] = items_cat
```

```python
rec_items_trend = []
seen = set()

for cat in top_categories:
    for it in cat_to_items.get(cat, []):
        if it not in seen:
            rec_items_trend.append(it)
            seen.add(it)

# đảm bảo đúng Hit@10
rec_items_trend = rec_items_trend[:K_HIT]

print("Trend items (Hit@10):", rec_items_trend)
```

```python
baseline_items = (
    trans_enriched
    .filter((pl.col("created_date") >= train_start) &
            (pl.col("created_date") <  train_end))
    .group_by("item_id")
    .count()
    .sort("count", descending=True)
    .head(K_HIT)
    .select("item_id")
)["item_id"].to_list()

print("Baseline items (Hit@10):", baseline_items)
```

```python
n_trend = int(round(K_HIT * HYBRID_ALPHA))   # 6
n_base  = K_HIT - n_trend                    # 4

hybrid_items = []
seen = set()

for it in rec_items_trend[:n_trend] + baseline_items:
    if it not in seen:
        hybrid_items.append(it)
        seen.add(it)
    if len(hybrid_items) == K_HIT:
        break

print("Hybrid items (Hit@10):", hybrid_items)
```

```python
def hitrate_at_k(test_df, rec_items):
    user_hit = (
        test_df
        .with_columns(pl.col("item_id").is_in(rec_items).alias("hit"))
        .group_by("customer_id")
        .agg(pl.col("hit").any().alias("user_hit"))
    )
    return user_hit.select(pl.col("user_hit").mean()).item()

print("Trend Hit@10   :", hitrate_at_k(test_trans_30d, rec_items_trend))
print("Baseline Hit@10:", hitrate_at_k(test_trans_30d, baseline_items))
print("Hybrid Hit@10  :", hitrate_at_k(test_trans_30d, hybrid_items))
```

```python
import pickle
import pandas as pd
import polars as pl

pkl_path = "final_groundtruth.pkl"

with open(pkl_path, "rb") as f:
    gt_obj = pickle.load(f)

# Case A: dict customer_id -> list[item_id]
if isinstance(gt_obj, dict):
    gt_pl = pl.DataFrame({
        "customer_id": list(gt_obj.keys()),
        "gt_items": list(gt_obj.values())
    })

# Case B/C: pandas DataFrame
elif isinstance(gt_obj, pd.DataFrame):
    gt_pl = pl.from_pandas(gt_obj)

# Case B/C: polars DataFrame
elif isinstance(gt_obj, pl.DataFrame):
    gt_pl = gt_obj

else:
    raise TypeError(f"Unsupported pkl type: {type(gt_obj)}")

# Chuẩn hóa về đúng schema: customer_id, gt_items(list)
if "gt_items" not in gt_pl.columns:
    if {"customer_id", "item_id"}.issubset(set(gt_pl.columns)):
        gt_pl = gt_pl.rename({
            "item_id": "gt_items"
            })
    else:
        raise ValueError(f"Ground truth must contain gt_items or (customer_id, item_id). Columns={gt_pl.columns}")

# Ép kiểu list + dọn null để tránh mean(None)
gt_pl = gt_pl.with_columns(
    pl.col("gt_items").cast(pl.List(pl.Utf8), strict=False).fill_null(pl.lit([]))
)

print("GT users:", gt_pl.height)
print("GT empty-list users:", gt_pl.filter(pl.col("gt_items").list.len() == 0).height)
```

```python
pred_202502 = gt_pl.with_columns(
    pl.lit(popular_item_list).alias("pred_items")
)

eval_df = pred_202502.with_columns(
    pl.col("pred_items")
      .list.set_intersection(pl.col("gt_items"))
      .list.len()
      .gt(0)
      .alias("hit")
)

hit_rate = eval_df.select(pl.col("hit").mean()).item()
print(f"HitRate@{TOP_K}:", hit_rate)
```

```python
from datetime import datetime

def add_months(dt: datetime, n: int) -> datetime:
    y = dt.year + (dt.month - 1 + n) // 12
    m = (dt.month - 1 + n) % 12 + 1
    return datetime(y, m, 1)
```

```python
from datetime import datetime

def get_cold_users_in_month(trans_m: pl.DataFrame, target_month: datetime):

    prev_start = add_months(target_month, -3)
    prev_end   = target_month   # không include target month

    users_t = (
        trans_m
        .filter(pl.col("month") == target_month)
        .select("customer_id")
        .unique()
    )

    users_prev = (
        trans_m
        .filter(
            (pl.col("month") >= prev_start) &
            (pl.col("month") <  prev_end)
        )
        .select("customer_id")
        .unique()
    )

    cold_users = users_t.join(
        users_prev,
        on="customer_id",
        how="anti"
    )

    return cold_users
```

```python
from datetime import datetime

start_month = datetime(2024, 4, 1)
end_month   = datetime(2025, 1, 1)
```

```python
cur = start_month
all_cold_items = []   # <-- PHẢI CÓ DÒNG NÀY
while cur <= end_month:

    print(f"Processing month {cur:%Y-%m}")

    cold_users = get_cold_users_in_month(trans_m, cur)

    if cold_users.height > 0:
        cold_tx = (
            trans_m
            .filter(pl.col("month") == cur)
            .join(cold_users, on="customer_id", how="inner")
        )
        all_cold_items.append(
            cold_tx
            .select("item_id")
        )

    cur = add_months(cur, 1)
```
