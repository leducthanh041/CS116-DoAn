```python
def read_parquet_by_type(train_path: str):
    # Lấy tất cả các file parquet trong thư mục
    files = [os.path.join(train_path, f) for f in os.listdir(train_path) if f.endswith('.parquet')]
    
    # Phân loại các file theo loại tên
    user_chunk_files = [file for file in files if 'user_chunk' in file]
    purchase_history_chunk_files = [file for file in files if 'purchase_history_daily_chunk' in file]
    item_chunk_files = [file for file in files if 'item_chunk' in file]
    
    # Đọc các file riêng biệt thành DataFrame
    user_chunk_df = pl.concat([pl.read_parquet(file) for file in user_chunk_files]) if user_chunk_files else None
    purchase_history_chunk_df = pl.concat([pl.read_parquet(file) for file in purchase_history_chunk_files]) if purchase_history_chunk_files else None
    item_chunk_df = pl.concat([pl.read_parquet(file) for file in item_chunk_files]) if item_chunk_files else None
    
    # Trả về một dictionary chứa các DataFrame
    return {
        "user_chunk": user_chunk_df,
        "purchase_history_chunk": purchase_history_chunk_df,
        "item_chunk": item_chunk_df
    }
```

### Tính Cramer's V giữa des_new và des với một số feature có thể liên quan

```python
import polars as pl
from math import sqrt

def cramers_v_fast_pl(df: pl.DataFrame, x: str, y: str):
    """
    Cramér's V (raw) giữa 2 cột phân loại x, y.
    - Chỉ drop NULL ở x,y (giữ nguyên dữ liệu gốc).
    - Không tạo cross-join r*c nên rất nhẹ.
    Trả về: (V, n, r, c)
    """
    d = df.select([x, y]).drop_nulls()
    n = d.height
    if n == 0:
        return float("nan"), 0, 0, 0

    # O_ij
    obs = d.group_by([x, y]).len().rename({"len": "obs"})
    # Tổng hàng & cột
    row = d.group_by(x).len().rename({"len": "row"})
    col = d.group_by(y).len().rename({"len": "col"})
    t = obs.join(row, on=x, how="left").join(col, on=y, how="left")

    # Chi-square: sum(O^2 / E) - N, với E = row_i * col_j / N
    term_sum = t.select(
        ((pl.col("obs") ** 2) / (pl.col("row") * pl.col("col") / n)).sum()
    ).to_series()[0]
    chi2 = term_sum - n
    r, c = d.select(pl.col(x).n_unique().alias("r"),
                    pl.col(y).n_unique().alias("c")).row(0)
    if r < 2 or c < 2:
        return float("nan"), n, r, c
    phi2 = chi2 / n
    V = sqrt(phi2 / min(r - 1, c - 1))
    return float(V), n, r, c

def compare_feature_pl(df: pl.DataFrame, feature: str) -> pl.DataFrame:
    rows = []
    for txt in ["description_new", "description"]:
        v, n, r, c = cramers_v_fast_pl(df, txt, feature)
        rows.append({"feature": feature, "text_col": txt,
                     "n_used": n, "x_levels": r, "y_levels": c,
                     "cramers_v": None if v != v else round(v, 3)})
    return pl.DataFrame(rows)
```

### Tính contain_ratio cho những dòng both present des_new và des

```python
import polars as pl
import re

# =======================
# 1) Tham số có thể chỉnh
# =======================
LEVEL = "word"   # "word" (theo từ) hoặc "char" (theo ký tự)
N     = 2        # n-gram: word bigram (2) / trigram (3) ; hoặc char-gram (ví dụ 5)

# =======================
# 2) Hàm tiện ích
# =======================
def tokenize(s: str):
    # Tách từ đơn giản, giữ unicode (phù hợp TV/TViệt)
    return re.findall(r"\w+", s.lower())

def ngrams(tokens, n=2):
    if len(tokens) < n: 
        return []
    return [" ".join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

def build_set(s: str, level="word", n=2):
    if not s:
        return set()
    if level == "char":
        s_norm = re.sub(r"\s+", " ", s.lower())
        if len(s_norm) < n:
            return set()
        return set(s_norm[i:i+n] for i in range(len(s_norm)-n+1))
    elif level == "word":
        toks = tokenize(s)
        if n == 1:
            return set(toks)
        else:
            return set(ngrams(toks, n))
    else:
        raise ValueError("LEVEL phải là 'word' hoặc 'char'.")

def contain_ratio_sets(A_set: set, B_set: set) -> float:
    return 0.0 if not A_set else len(A_set & B_set) / len(A_set)

def contain_pair(a: str, b: str, level="word", n=2):
    """
    Trả về (recall=A|B, precision=B|A, F1, sizeA, sizeB)
    """
    A = build_set(a, level=level, n=n)
    B = build_set(b, level=level, n=n)
    r = contain_ratio_sets(A, B)  # A nằm trong B bao nhiêu %
    p = contain_ratio_sets(B, A)  # B nằm trong A bao nhiêu %
    f1 = 0.0 if (p + r) == 0 else 2 * p * r / (p + r)
    return (r, p, f1, len(A), len(B))

# ===========================================
# 3) Mask both_present & báo cáo tình trạng null
# ===========================================
mask_both = (
    pl.col("description").is_not_null() & pl.col("description_new").is_not_null() &
    (pl.col("description").str.len_chars() > 0) &
    (pl.col("description_new").str.len_chars() > 0)
)

null_report = item_df.select([
    (pl.col("description").is_null() & pl.col("description_new").is_null()).mean().alias("both_null_frac"),
    (pl.col("description").is_null() & pl.col("description_new").is_not_null()).mean().alias("desc_null_only_frac"),
    (pl.col("description").is_not_null() & pl.col("description_new").is_null()).mean().alias("desc_new_null_only_frac"),
    mask_both.mean().alias("both_present_frac"),
])
print("=== Null report (fraction) ===")
print(null_report)

# ===========================================
# 4) Tính contain trên tập both_present
# ===========================================
both_df = item_df.filter(mask_both)

# Thêm 3 cột: recall(A|B), precision(B|A), F1
res = both_df.with_columns([
    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[0]
    ).alias("contain_desc_in_descnew"),  # recall: A|B

    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[1]
    ).alias("contain_descnew_in_desc"),  # precision: B|A

    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[2]
    ).alias("contain_f1"),
])

# Tóm tắt nhanh
summary = res.select([
    pl.len().alias("n_both_present"),
    pl.col("contain_desc_in_descnew").mean().alias("mean_contain(A|B)"),
    pl.col("contain_descnew_in_desc").mean().alias("mean_contain(B|A)"),
    pl.col("contain_f1").mean().alias("mean_contain_F1"),
    pl.col("contain_desc_in_descnew").quantile(0.5).alias("median_contain(A|B)"),
    pl.col("contain_desc_in_descnew").quantile(0.9).alias("p90_contain(A|B)"),
])
print("=== Containment summary (LEVEL={}, N={}) ===".format(LEVEL, N))
print(summary)

# (Tuỳ chọn) xem vài dòng có F1 cao để sanity-check
print(res.select(["description", "description_new", "contain_desc_in_descnew", "contain_descnew_in_desc", "contain_f1"])
        .sort("contain_f1", descending=True)
        .head(10))
```

Ở trên chưa bỏ Không xác định -> sai

```python
import polars as pl
import re

# ===== 1) Cấu hình =====
LEVEL = "word"   # "word" hoặc "char"
N = 2            # n-gram: 1/2/3... ; ví dụ word-bigram = 2

# Các giá trị được coi là "thiếu"
PLACEHOLDERS = [
    "", "-", "--",
    "không xác định", "khong xac dinh", "không xac dinh",
    "không rõ", "chưa rõ", "không có thông tin",
    "n/a", "na", "none", "null", "unk", "unknown"
]

# ===== 2) Hàm tiện ích =====
def tokenize(s: str):
    return re.findall(r"\w+", s.lower())

def ngrams(tokens, n=2):
    if len(tokens) < n: 
        return []
    return [" ".join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

def build_set(s: str, level="word", n=2):
    if not s:
        return set()
    if level == "char":
        s_norm = re.sub(r"\s+", " ", s.lower()).strip()
        if len(s_norm) < n:
            return set()
        return set(s_norm[i:i+n] for i in range(len(s_norm)-n+1))
    else:
        toks = tokenize(s)
        return set(toks) if n == 1 else set(ngrams(toks, n))

def contain_ratio_sets(A_set: set, B_set: set) -> float:
    return 0.0 if not A_set else len(A_set & B_set) / len(A_set)

def contain_pair(a: str, b: str, level="word", n=2):
    A = build_set(a, level=level, n=n)
    B = build_set(b, level=level, n=n)
    r = contain_ratio_sets(A, B)  # Contain(A|B)
    p = contain_ratio_sets(B, A)  # Contain(B|A)
    f1 = 0.0 if (p + r) == 0 else 2 * p * r / (p + r)
    return (r, p, f1, len(A), len(B))

# Chuẩn hoá text để so đối chiếu placeholder
def norm_expr(colname: str):
    return (
        pl.col(colname).cast(pl.Utf8)
        .str.to_lowercase()
        .str.replace_all(r"\s+", " ")
        .str.strip_chars(" .,-:;—–_/\\|\"'`")  # bỏ dấu câu ở đầu/cuối
        .str.strip_chars()                     # bỏ khoảng trắng thừa
    )

def is_missing(colname: str):
    n = norm_expr(colname)
    return pl.col(colname).is_null() | n.is_in(PLACEHOLDERS)

# ===== 3) Mask both_present với định nghĩa "thiếu" mới =====
miss_desc     = is_missing("description")
miss_desc_new = is_missing("description_new")
both_present  = (~miss_desc) & (~miss_desc_new)

# Báo cáo thiếu dữ liệu (fraction)
null_report = item_df.select([
    (miss_desc & miss_desc_new).mean().alias("both_null_frac"),
    (miss_desc & (~miss_desc_new)).mean().alias("desc_null_only_frac"),
    ((~miss_desc) & miss_desc_new).mean().alias("desc_new_null_only_frac"),
    both_present.mean().alias("both_present_frac"),
])
print("=== Null report (fraction) — placeholders treated as null ===")
print(null_report)

# ===== 4) Tính containment chỉ trên both_present =====
both_df = item_df.filter(both_present)

res = both_df.with_columns([
    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[0]
    ).alias("contain_desc_in_descnew"),   # recall: A|B

    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[1]
    ).alias("contain_descnew_in_desc"),   # precision: B|A

    pl.struct(["description", "description_new"]).map_elements(
        lambda row: contain_pair(row["description"], row["description_new"], level=LEVEL, n=N)[2]
    ).alias("contain_f1"),
])

summary = res.select([
    pl.len().alias("n_both_present"),
    pl.col("contain_desc_in_descnew").mean().alias("mean_contain(A|B)"),
    pl.col("contain_descnew_in_desc").mean().alias("mean_contain(B|A)"),
    pl.col("contain_f1").mean().alias("mean_contain_F1"),
    pl.col("contain_desc_in_descnew").quantile(0.5).alias("median_contain(A|B)"),
    pl.col("contain_desc_in_descnew").quantile(0.9).alias("p90_contain(A|B)"),
])
print(f"=== Containment summary (LEVEL={LEVEL}, N={N}) — after placeholder filtering ===")
print(summary)

# (Tuỳ chọn) kiểm tra vài dòng F1 cao
print(res.select(["description", "description_new", "contain_desc_in_descnew", "contain_descnew_in_desc", "contain_f1"])
        .sort("contain_f1", descending=True)
        .head(10))
```

### Kiểm tra ngoài "Không xác định" và null thì còn giá trị nào có nghĩa là missing không

```python
df = item_df

def audit_missing_col(df: pl.DataFrame, col: str) -> pl.DataFrame:
    c = pl.col(col)
    norm = (
        c.cast(pl.Utf8)
         .str.replace_all('"', '')        # bỏ ngoặc kép
         .str.replace_all('“|”', '')      # bỏ ngoặc kép unicode nếu có
         .str.to_lowercase()
         .str.strip_chars()
    )
    # các biến thể thường gặp cần rà soát
    candidates = ["không xác định", "khong xac dinh", "null", "none", "na", "n/a", ""]
    return (
        df.select([
            pl.count().alias("total"),
            c.is_null().sum().alias("null_rows"),
            (norm == "không xác định").sum().alias("khong_xac_dinh"),
            (norm == "khong xac dinh").sum().alias("khong_xac_dinh_khong_dau"),
            ((norm == "") & c.is_not_null()).sum().alias("empty_after_trim"),
            (norm.is_in(["null","none","na","n/a"]) & c.is_not_null()).sum().alias("other_placeholders"),
            # chuỗi chỉ toàn dấu ngăn cách (nếu dataset có kiểu này)
            (
              c.is_not_null() &
              (c.str.replace_all(r"[ \-\.\_/|:;]+", "").str.len_chars() == 0)
            ).sum().alias("only_separators")
        ])
        .with_columns(col_name = pl.lit(col))
        .select(["col_name","total","null_rows","khong_xac_dinh","khong_xac_dinh_khong_dau",
                 "empty_after_trim","other_placeholders","only_separators"])
    )

# Chạy audit cho 2 cột
result = pl.concat([
    audit_missing_col(df, "description_new"),
    audit_missing_col(df, "description"),
])
print(result)

# (tùy chọn) Liệt kê nhanh top 10 giá trị "nghi ngờ placeholder"
for col in ["description_new","description"]:
    c = pl.col(col).cast(pl.Utf8)
    norm = (c.str.replace_all('"','').str.replace_all('“|”','')
              .str.to_lowercase().str.strip_chars())
    suspicious = df.select([norm.alias("norm")]) \
                   .group_by("norm").len() \
                   .filter(pl.col("norm").is_in(["không xác định","khong xac dinh","null","none","na","n/a",""])) \
                   .sort("len", descending=True)
    print(f"\n{col} — placeholders (normalized):")
    print(suspicious)
```

### Gộp 2 cột description và description_new

- Sau khi đã xác định các giá trị thiếu chỉ gồm null hoặc "Không xác định" thì chỉ cần kiểm tra missing(x) = is_null(x) OR lower(strip(x)) == "Không xác định" 
- Để nhận diện thiếu, tránh copy nhầm "Không xác định"

**Logic gộp**
1. Nếu des_new không missing => lấy des_new
2. Nếu des_new missing nhưng des không missing => lấy des 
3. Nếu cả hai missing => Để NULL
4. Thêm cột description_source = new / old_as_fill / missing để audit

```python
df = item_df

# 1) Định nghĩa điều kiện thiếu (tối giản)
def is_missing(col: str) -> pl.Expr:
    c = pl.col(col)
    return c.is_null() | (
        c.cast(pl.Utf8).str.to_lowercase().str.strip_chars() == "không xác định"
    )

new_miss = is_missing("description_new")
old_miss = is_missing("description")

# 2) Gộp và đánh dấu nguồn
df = df.with_columns([
    pl.when(~new_miss).then(pl.col("description_new"))
      .when(~old_miss).then(pl.col("description"))
      .otherwise(None)
      .alias("description_merged"),

    pl.when(~new_miss).then(pl.lit("new"))
     .when(~old_miss).then(pl.lit("old_as_fill"))
     .otherwise(pl.lit("missing"))
     .alias("description_source")
])

# 3) Báo cáo: filled_from_old & still_missing (+ %)
report = (
    df.select(
        pl.count().alias("total"),
        (pl.col("description_source") == "old_as_fill").sum().alias("filled_from_old"),
        (pl.col("description_source") == "missing").sum().alias("still_missing"),
        (pl.col("description_source") == "new").sum().alias("from_new")
    )
    .with_columns(
        filled_from_old_pct = (pl.col("filled_from_old") / pl.col("total") * 100).round(2),
        still_missing_pct   = (pl.col("still_missing")   / pl.col("total") * 100).round(2),
    )
)

print(">>> Merge report")
print(report)

# 4) (Tuỳ chọn) xem nhanh vài dòng được lấp từ description cũ
print("\n>>> Sample rows (old_as_fill):")
print(
    df.filter(pl.col("description_source") == "old_as_fill")
      .select(["description_new", "description", "description_merged"])
      .head(5)
)

# Gán lại vào item_df nếu bạn muốn tiếp tục dùng ở các cell sau
item_df = df
```

```python
# ============================================================
# RELATION & COVERAGE between item_df["size"] and ["age_group"]
# ============================================================
import polars as pl
import math, re

# -------------------------------
# 0) Chuẩn hoá giá trị null/unknown
# -------------------------------
NULLISH = {"", "không xác định", "khong xac dinh", "unknown", "null", "none", "nan", "na", "n/a"}

def _clean_cat(x):
    if x is None:
        return None
    s = str(x).strip()
    return None if s.lower() in NULLISH else s

df2 = item_df.with_columns([
    pl.col("size").map_elements(_clean_cat, return_dtype=pl.Utf8).alias("size_c"),
    pl.col("age_group").map_elements(_clean_cat, return_dtype=pl.Utf8).alias("age_c"),
])

pair = df2.select("size_c", "age_c").drop_nulls()
N = pair.height
print(f"[INFO] Rows used (both non-null): {N}")
if N == 0:
    raise ValueError("Không đủ dữ liệu hợp lệ để tính các chỉ số (N == 0).")

# -------------------------------
# 1) Contingency table
# -------------------------------
ct = pair.group_by(["size_c", "age_c"]).len().rename({"len": "n"})
row_tot = ct.group_by("size_c").agg(pl.col("n").sum().alias("row_total"))
col_tot = ct.group_by("age_c").agg(pl.col("n").sum().alias("col_total"))

expected_expr = (pl.col("row_total") * pl.col("col_total") / pl.lit(N))
ct2 = (
    ct.join(row_tot, on="size_c")
      .join(col_tot, on="age_c")
      .with_columns([
          ((pl.col("n") - expected_expr) ** 2 / expected_expr).alias("chi2_term"),
          (pl.col("n") / pl.lit(N)).alias("p_xy"),
      ])
)

# -------------------------------
# 2) Cramér’s V (bias-corrected)
# -------------------------------
chi2 = ct2.select(pl.col("chi2_term").sum()).item()
r, c = row_tot.height, col_tot.height
cramers_v = 0.0
if r >= 2 and c >= 2:
    phi2 = chi2 / N
    phi2corr = max(0.0, phi2 - ((c - 1) * (r - 1)) / (N - 1))
    rcorr = r - 1 - (r - 1) ** 2 / (N - 1)
    ccorr = c - 1 - (c - 1) ** 2 / (N - 1)
    denom = max(1.0, min(rcorr, ccorr))
    cramers_v = math.sqrt(phi2corr / denom) if denom > 0 else 0.0

# -------------------------------
# 3) Mutual Information, NMI, Theil’s U
# -------------------------------
row_probs = row_tot.with_columns((pl.col("row_total") / pl.lit(N)).alias("p_x"))
col_probs = col_tot.with_columns((pl.col("col_total") / pl.lit(N)).alias("p_y"))

ct3 = (
    ct2.join(row_probs.select(["size_c", "p_x"]), on="size_c")
       .join(col_probs.select(["age_c", "p_y"]), on="age_c")
       .with_columns(
           pl.when(pl.col("p_xy") > 0)
             .then(pl.col("p_xy") * (pl.col("p_xy") / (pl.col("p_x") * pl.col("p_y"))).log())
             .otherwise(0.0)
             .alias("mi_term")
       )
)

mi = ct3.select(pl.col("mi_term").sum()).item()
Hx = -row_probs.select((pl.col("p_x") * pl.col("p_x").log()).sum()).item()  # entropy(size)
Hy = -col_probs.select((pl.col("p_y") * pl.col("p_y").log()).sum()).item()  # entropy(age_group)

nmi = mi / math.sqrt(Hx * Hy) if Hx > 0 and Hy > 0 else 0.0
U_age_given_size = mi / Hy if Hy > 0 else 0.0    # Theil's U(Y|X)
U_size_given_age = mi / Hx if Hx > 0 else 0.0    # Theil's U(X|Y)

# -------------------------------
# 4) Mode-coverage hai chiều (per-group mode)
#    Dùng rank over group để lấy phần tử n lớn nhất per group (ổn định)
# -------------------------------
# size -> age_group
top_by_size = (
    ct.with_columns(pl.col("n").rank("dense", descending=True).over("size_c").alias("rk"))
      .filter(pl.col("rk") == 1)
      .select(["size_c", "age_c", "n"])
      .join(row_tot, on="size_c")
      .with_columns((pl.col("n") / pl.col("row_total")).alias("share_in_size"))
      .sort("share_in_size", descending=True)
)
acc_size_to_age = top_by_size.select(pl.col("n").sum() / pl.lit(N)).item()

# age_group -> size
top_by_age = (
    ct.with_columns(pl.col("n").rank("dense", descending=True).over("age_c").alias("rk"))
      .filter(pl.col("rk") == 1)
      .select(["age_c", "size_c", "n"])
      .join(col_tot, on="age_c")
      .with_columns((pl.col("n") / pl.col("col_total")).alias("share_in_age"))
      .sort("share_in_age", descending=True)
)
acc_age_to_size = top_by_age.select(pl.col("n").sum() / pl.lit(N)).item()

# -------------------------------
# 5) Metrics table (ép dtype rõ ràng để tránh lỗi int/float)
# -------------------------------
metrics = pl.DataFrame({
    "metric": pl.Series(
        "metric",
        [
            "N",
            "Cramers_V",
            "NMI",
            "Theils_U(age|size)",
            "Theils_U(size|age)",
            "ModeCoverage(size→age)",
            "ModeCoverage(age→size)",
        ],
        dtype=pl.Utf8,
    ),
    "value": pl.Series(
        "value",
        [
            float(N),
            float(cramers_v),
            float(nmi),
            float(U_age_given_size),
            float(U_size_given_age),
            float(acc_size_to_age),
            float(acc_age_to_size),
        ],
        dtype=pl.Float64,
    ),
})

print("\n===== METRICS =====")
print(metrics)

print("\n===== TOP MAP: size → age_group (per-size mode) =====")
print(top_by_size.head(30))   # đổi head(None) nếu muốn in tất cả

print("\n===== TOP MAP: age_group → size (per-age mode) =====")
print(top_by_age.head(30))    # đổi head(None) nếu muốn in tất cả

# -------------------------------
# 6) (Optional) So khớp token tuổi giữa 2 cột
# -------------------------------
DO_TOKEN_MATCH = True

def _age_token(s: str | None):
    if s is None:
        return None
    s = s.lower().strip()
    s = (
        s.replace("tháng", "m").replace("thang", "m")
         .replace("tuổi", "y").replace("tuoi", "y")
         .replace("–", "-").replace("—", "-").replace("−", "-")
         .replace(" ", "")
    )
    # range month/year
    if re.fullmatch(r"\d+-\d+m", s): return s     # ví dụ: 6-9m
    if re.fullmatch(r"\d+-\d+y", s): return s     # ví dụ: 1-3y
    # single month/year
    if re.fullmatch(r"\d+m", s): return s         # ví dụ: 9m
    if re.fullmatch(r"\d+y", s): return s         # ví dụ: 2y
    # prefixes
    if s.startswith("từ") or s.startswith("tu"):
        return "tu" + re.sub(r"[^\dy]", "", s.replace("từ", "tu"))
    if s.startswith("trên") or s.startswith("tren"):
        return "tren" + re.sub(r"[^\dy]", "", s.replace("trên", "tren"))
    return None

if DO_TOKEN_MATCH:
    df_tok = df2.with_columns([
        pl.col("size_c").map_elements(_age_token, return_dtype=pl.Utf8).alias("size_token"),
        pl.col("age_c").map_elements(_age_token, return_dtype=pl.Utf8).alias("age_token"),
    ])
    both = df_tok.filter(pl.col("size_token").is_not_null() & pl.col("age_token").is_not_null())
    exact_match_rate = both.select(
        (pl.col("size_token") == pl.col("age_token")).cast(pl.Float64).mean().alias("rate")
    ).item()
    print(f"\n[OPTIONAL] Exact-match rate of normalized age tokens: {exact_match_rate:.3f} "
          f"(on {both.height} tokenized rows)")
```

```python
import polars as pl
import re, json

# ── Chuẩn hoá chuỗi để parse ──────────────────────────────────────────────
def _norm_age(s: str) -> str:
    s = s.lower().strip()
    s = (s.replace("tháng","m").replace("thang","m")
           .replace("tuổi","y").replace("tuoi","y")
           .replace("từ","tu").replace("trên","tren")
           .replace("–","-").replace("—","-").replace("−","-")
           .replace(" ", ""))
    # toddler 'T' -> 'y' (2T = 2y)
    s = re.sub(r"(?<=\d)t\b", "y", s)
    # gom newborn
    s = (s.replace("sơsinh","nb")
           .replace("sosinh","nb")
           .replace("newborn","nb"))
    return s

# ── Parse về (lo, hi) theo THÁNG; cover cả "0-5Y", "0-4Y", ... ────────────
def _parse_to_months(val: str | None):
    if val is None:
        return (None, None)

    raw = str(val).strip()

    # list-string JSON: ["Từ 0M","Từ 3M",...]
    if raw.lstrip().startswith("["):
        try:
            items = json.loads(raw)
            lows, highs = [], []
            for it in items:
                lo, hi = _parse_to_months(it)
                if lo is not None: lows.append(lo)
                if hi is not None: highs.append(hi)
            if lows or highs:
                return (min(lows) if lows else None, max(highs) if highs else None)
        except Exception:
            pass  # rơi xuống parse thường

    s = _norm_age(raw)

    # giữ nguyên "Mẹ"
    if s in {"me","mẹ"}:
        return ("Mẹ", "Mẹ")  # dùng sentinel để lát nữa trả về "Mẹ"

    # newborn
    if s == "nb":
        return (0, 3)

    # 1) RANGE với đơn vị tuỳ chọn ở MỖI VẾ
    #    - bắt được: 0-5y, 6m-12m, 12m-4y, 1y-3y, v.v...
    m = re.fullmatch(r"(\d+)(m|y)?-(\d+)(m|y)?", s)
    if m:
        a, u1, b, u2 = m.groups()
        if u1 is None and u2 is None:
            pass  # không đoán bừa nếu cả hai đều thiếu
        else:
            # nếu thiếu đơn vị ở 1 vế, kế thừa từ vế còn lại
            if u1 is None: u1 = u2
            if u2 is None: u2 = u1
            a = int(a) * (1 if u1 == "m" else 12)
            b = int(b) * (1 if u2 == "m" else 12)
            if a > b: a, b = b, a
            return (a, b)

    # 2) SINGLE: 9m, 2y
    m = re.fullmatch(r"(\d+)(m|y)", s)
    if m:
        a, u = m.groups()
        a = int(a) * (1 if u == "m" else 12)
        return (a, a)

    # 3) THRESHOLD: tu6m, tren2y
    m = re.fullmatch(r"(tu|tren)(\d+)(m|y)", s)
    if m:
        v = int(m.group(2)) * (1 if m.group(3) == "m" else 12)
        return (v, None)

    return (None, None)

# ── Map (lo,hi) -> primary bucket ─────────────────────────────────────────
def _bucket_from_months(lo, hi):
    if lo == "Mẹ" and hi == "Mẹ":
        return "Mẹ"
    if lo is None and hi is None:
        return None
    mid = lo if (hi is None or isinstance(lo, str)) else (lo + hi) / 2.0
    if isinstance(mid, str):  # đề phòng "Mẹ"
        return "Mẹ"
    if mid < 6:   return "0-6M"
    if mid < 12:  return "6-12M"
    if mid < 24:  return "1-2Y"
    if mid < 48:  return "2-4Y"
    return ">4Y"

def age_to_primary_bucket(val):
    lo, hi = _parse_to_months(val)
    return _bucket_from_months(lo, hi)

# ── Tạo cột bucket ────────────────────────────────────────────────────────
item_df = item_df.with_columns(
    pl.col("age_group").map_elements(age_to_primary_bucket, return_dtype=pl.Utf8).alias("age_bucket_primary")
)

# (tuỳ chọn) xem phân bố
print(item_df.select(pl.col("age_bucket_primary").value_counts()))
```

```python
import polars as pl
import re, json
from functools import lru_cache

# =========================
# 0) Tham số
# =========================
USE_LLM = False           # Bật True nếu bạn đã điền call_llm_api bên dưới
LLM_THRESHOLD = 0.90      # Ngưỡng tin cậy tối thiểu để chấp nhận kết quả LLM
MAX_MONTHS = 216          # Giới hạn biên (18 năm), điều chỉnh nếu cần

# =========================
# 1) Chuẩn hoá/bucket hoá AGE (tái dùng cho size dạng tuổi)
# =========================
def _norm_age(s: str) -> str:
    s = s.lower().strip()
    s = (s.replace("tháng","m").replace("thang","m")
           .replace("tuổi","y").replace("tuoi","y")
           .replace("từ","tu").replace("trên","tren")
           .replace("–","-").replace("—","-").replace("−","-")
           .replace(" ", ""))
    # toddler 'T' -> 'y' (2T=2y)
    s = re.sub(r"(?<=\d)t\b", "y", s)
    # newborn
    s = (s.replace("sơsinh","nb")
           .replace("sosinh","nb")
           .replace("newborn","nb"))
    return s

def parse_age_text_to_months(val: str | None):
    """Parse chuỗi tuổi (size dạng tuổi HOẶC age_group) → (lo_m, hi_m).
       Cover: range trộn đơn vị + thiếu đơn vị 1 vế, single, threshold, list-string JSON, NB.
    """
    if val is None:
        return (None, None)

    raw = str(val).strip()

    # list-string JSON: ["Từ 0M","Từ 3M",...]
    if raw.lstrip().startswith("["):
        try:
            items = json.loads(raw)
            lows, highs = [], []
            for it in items:
                lo, hi = parse_age_text_to_months(it)
                if lo is not None: lows.append(lo)
                if hi is not None: highs.append(hi)
            if lows or highs:
                return (min(lows) if lows else None, max(highs) if highs else None)
        except Exception:
            pass  # rơi xuống parse thường

    s = _norm_age(raw)

    # giữ "mẹ": không phải tuổi
    if s in {"me","mẹ"}:
        return (None, None)

    # newborn
    if s == "nb":
        return (0, 3)

    # RANGE với đơn vị tuỳ chọn ở hai vế (0-5y, 6m-12m, 12m-4y, 1y-3y, ...)
    m = re.fullmatch(r"(\d+)(m|y)?-(\d+)(m|y)?", s)
    if m:
        a, u1, b, u2 = m.groups()
        if u1 is None and u2 is None:
            pass  # không đoán khi cả hai thiếu
        else:
            if u1 is None: u1 = u2
            if u2 is None: u2 = u1
            a = int(a) * (1 if u1 == "m" else 12)
            b = int(b) * (1 if u2 == "m" else 12)
            if a > b: a, b = b, a
            return (a, b)

    # SINGLE: 9m, 2y
    m = re.fullmatch(r"(\d+)(m|y)", s)
    if m:
        a, u = m.groups()
        a = int(a) * (1 if u == "m" else 12)
        return (a, a)

    # THRESHOLD: tu6m, tren2y
    m = re.fullmatch(r"(tu|tren)(\d+)(m|y)", s)
    if m:
        v = int(m.group(2)) * (1 if m.group(3) == "m" else 12)
        return (v, None)

    return (None, None)

def primary_bucket_from_months(lo: int | None, hi: int | None) -> str | None:
    """Mid-based primary bucket (biên 6/12/24/48 dùng '<' để rơi sang bucket cao hơn)."""
    if lo is None and hi is None:
        return None
    mid = lo if hi is None else (lo + hi) / 2.0
    if mid < 6:   return "0-6M"
    if mid < 12:  return "6-12M"
    if mid < 24:  return "1-2Y"
    if mid < 48:  return "2-4Y"
    return ">4Y"

# =========================
# 2) RULES: nhận diện size dạng tuổi
# =========================
AGE_KEYWORDS = None  # không dùng keyword; rely hoàn toàn vào regex tuổi bên trên

def size_rules_extract_months(size_val: str | None):
    """Nếu size là tuổi -> (lo_m, hi_m), else (None, None)."""
    if size_val is None:
        return (None, None)
    lo, hi = parse_age_text_to_months(size_val)
    # validate biên
    if lo is None and hi is None:
        return (None, None)
    # chặn các case vô lý
    l = lo if lo is not None else 0
    h = hi if hi is not None else MAX_MONTHS
    if l < 0 or h < 0 or h > MAX_MONTHS or (hi is not None and lo is not None and lo > hi):
        return (None, None)
    return (lo, hi)

# =========================
# 3) (TÙY CHỌN) LLM/API extraction
# =========================
def call_llm_api(size_text: str) -> dict | None:
    """
    >>> TRÁI TIM LOGIC Ở ĐÂY <<<
    Trả về dict như: {"type":"age","lo_m":6,"hi_m":12,"confidence":0.93}
    hoặc None nếu không chắc.

    Bạn có thể:
    - Dùng OpenAI: gpt-4o-mini / structured JSON
    - Dùng local Ollama: llama3/qwen; prompt yêu cầu JSON thuần
    - Dùng bất kỳ endpoint nào bạn có

    Ở đây mình để khung trống (return None).
    """
    return None  # <-- điền code gọi LLM của bạn ở đây

def size_llm_extract_months(size_val: str | None, threshold: float = LLM_THRESHOLD):
    if not size_val:
        return (None, None, None)  # lo, hi, conf
    resp = call_llm_api(str(size_val))
    if not resp:
        return (None, None, None)
    try:
        if resp.get("type") != "age":
            return (None, None, resp.get("confidence"))
        lo = resp.get("lo_m"); hi = resp.get("hi_m"); conf = float(resp.get("confidence", 0))
        # validate
        if conf < threshold: 
            return (None, None, conf)
        if lo is None and hi is None:
            return (None, None, conf)
        l = 0 if lo is None else int(lo)
        h = MAX_MONTHS if hi is None else int(hi)
        if l < 0 or h < 0 or h > MAX_MONTHS or (lo is not None and hi is not None and lo > hi):
            return (None, None, conf)
        return (lo, hi, conf)
    except Exception:
        return (None, None, None)

# =========================
# 4) Áp dụng trên các giá trị size UNIQUE (cache + join lại)
# =========================
# Lấy unique size (bạn có thể thay "size" bằng cột bạn dùng)
size_unique = (
    item_df.select(pl.col("size").cast(pl.Utf8))
           .unique()
           .drop_nulls()
)

# Tạo mapping qua rules trước
size_map_rows = []
for s in size_unique.get_column("size").to_list():
    lo, hi = size_rules_extract_months(s)
    source = None
    conf = None
    if lo is not None or hi is not None:
        source = "size_rule"
    elif USE_LLM:
        lo, hi, conf = size_llm_extract_months(s, threshold=LLM_THRESHOLD)
        if lo is not None or hi is not None:
            source = "size_llm"
    # bucket
    bucket = primary_bucket_from_months(lo, hi) if (lo is not None or hi is not None) else None
    size_map_rows.append( (s, lo, hi, bucket, source, conf) )

size_map = pl.DataFrame(size_map_rows, schema={
    "size": pl.Utf8,
    "size_lo_m": pl.Int64,
    "size_hi_m": pl.Int64,
    "size_age_bucket": pl.Utf8,
    "age_fill_source": pl.Utf8,
    "age_llm_conf": pl.Float64,
})

# Join mapping vào item_df
df_with_size_age = (
    item_df.join(size_map, on="size", how="left")
)

# Điền bổ sung vào age_bucket_primary nếu đang null
df_filled = df_with_size_age.with_columns(
    pl.when(pl.col("age_bucket_primary").is_null() & pl.col("size_age_bucket").is_not_null())
      .then(pl.col("size_age_bucket"))
      .otherwise(pl.col("age_bucket_primary"))
      .alias("age_bucket_primary_filled")
)

# Báo cáo nhanh coverage tăng thêm bao nhiêu
coverage_before = df_with_size_age.select(pl.col("age_bucket_primary").is_not_null().sum()).item()
coverage_after  = df_filled.select(pl.col("age_bucket_primary_filled").is_not_null().sum()).item()
total_rows      = df_filled.height
print({
    "total_rows": total_rows,
    "coverage_before": coverage_before,
    "coverage_after": coverage_after,
    "delta": coverage_after - coverage_before
})
```

```python
# === Cell: Trích tuổi từ description_merged và điền bổ sung (đÃ FIX "0-6M") ===
import polars as pl

df = item_df  # dùng DF hiện tại

# -- helper: map mid(tháng) -> nhãn bucket (dùng pl.lit cho LITERAL) --
def bucket_from_mid_expr(mid: pl.Expr) -> pl.Expr:
    return (
        pl.when(mid.is_null()).then(pl.lit(None))
         .when(mid < 6).then(pl.lit("0-6M"))
         .when(mid < 12).then(pl.lit("6-12M"))
         .when(mid < 24).then(pl.lit("1-2Y"))
         .when(mid < 48).then(pl.lit("2-4Y"))
         .otherwise(pl.lit(">4Y"))
    )

# (an toàn) nếu có cột bucket cũ dùng en-dash, đưa về ASCII hyphen
if "age_bucket_primary_filled" in df.columns:
    df = df.with_columns(
        pl.col("age_bucket_primary_filled").cast(pl.Utf8)
          .str.replace_all("–|—", "-")
          .alias("age_bucket_primary_filled")
    )

# nếu chưa có cột primary, tạo từ age_month_mid (nếu có), else để None
if "age_bucket_primary_filled" not in df.columns:
    df = df.with_columns(
        bucket_from_mid_expr(pl.col("age_month_mid")).alias("age_bucket_primary_filled")
        if "age_month_mid" in df.columns else pl.lit(None).alias("age_bucket_primary_filled")
    )

# 1) chuẩn hoá mô tả để parse (không đổi giá trị gốc)
df = df.with_columns(
    pl.col("description_merged").cast(pl.Utf8)
      .str.to_lowercase()
      .str.replace_all('"', '').str.replace_all('“|”', '')
      .str.replace_all(r"\s+", " ").str.strip_chars()
      .alias("_desc_norm")
)

# 2) trích tuổi (tháng) từ mô tả
d_range_m_lo = pl.col("_desc_norm").str.extract(r'(\d{1,2})\s*[-–—]\s*(\d{1,2})\s*(?:m|mo|tháng)\b', 1).cast(pl.Int64)
d_range_m_hi = pl.col("_desc_norm").str.extract(r'(\d{1,2})\s*[-–—]\s*(\d{1,2})\s*(?:m|mo|tháng)\b', 2).cast(pl.Int64)
d_range_y_lo = pl.col("_desc_norm").str.extract(r'(\d{1,2})\s*[-–—]\s*(\d{1,2})\s*(?:y|yr|years|tuổi|năm)\b', 1).cast(pl.Int64) * 12
d_range_y_hi = pl.col("_desc_norm").str.extract(r'(\d{1,2})\s*[-–—]\s*(\d{1,2})\s*(?:y|yr|years|tuổi|năm)\b', 2).cast(pl.Int64) * 12
d_lb_m       = pl.col("_desc_norm").str.extract(r'(?:từ|>=|trên)\s*(\d{1,2})\s*(?:m|mo|tháng)\b', 1).cast(pl.Int64)
d_lb_y       = pl.col("_desc_norm").str.extract(r'(?:từ|>=|trên)\s*(\d{1,2})\s*(?:y|yr|years|tuổi|năm)\b', 1).cast(pl.Int64) * 12
d_single_m   = pl.col("_desc_norm").str.extract(r'\b(\d{1,2})\s*(?:m|mo|tháng)\b', 1).cast(pl.Int64)
d_single_y   = pl.col("_desc_norm").str.extract(r'\b(\d{1,2})\s*(?:y|yr|years|tuổi|năm)\b', 1).cast(pl.Int64) * 12
d_nb_lo      = pl.when(pl.col("_desc_norm").str.contains(r'\b(nb|newborn|sơ sinh)\b')).then(pl.lit(0)).otherwise(None)
d_nb_hi      = pl.when(pl.col("_desc_norm").str.contains(r'\b(nb|newborn|sơ sinh)\b')).then(pl.lit(3)).otherwise(None)

df = df.with_columns([
    pl.coalesce([d_range_m_lo, d_range_y_lo, d_nb_lo, d_lb_m, d_lb_y, d_single_m, d_single_y]).alias("desc_age_month_lo"),
    pl.coalesce([d_range_m_hi, d_range_y_hi, d_nb_hi]).alias("desc_age_month_hi"),
]).with_columns([
    pl.when(pl.col("desc_age_month_lo").is_not_null() & pl.col("desc_age_month_hi").is_not_null())
      .then((pl.col("desc_age_month_lo") + pl.col("desc_age_month_hi")) / 2)
      .when(pl.col("desc_age_month_lo").is_not_null())
      .then(pl.col("desc_age_month_lo").cast(pl.Float64))
      .otherwise(None)
      .alias("desc_age_month_mid"),
])

# 3) bucket mô tả theo chuẩn ASCII (dùng pl.lit trong bucket_from_mid_expr)
df = df.with_columns(
    bucket_from_mid_expr(pl.col("desc_age_month_mid")).alias("desc_age_bucket")
)

# 4) confidence đơn giản
df = df.with_columns(
    pl.when(pl.col("desc_age_month_lo").is_not_null() & pl.col("desc_age_month_hi").is_not_null()).then(pl.lit(0.95))
     .when(pl.col("desc_age_month_lo").is_not_null() & pl.col("desc_age_month_hi").is_null()).then(pl.lit(0.85))
     .otherwise(pl.lit(None))
     .alias("desc_age_confidence")
)

# 5) điền bổ sung vào tuổi chính (coalesce ưu tiên primary)
df = df.with_columns([
    pl.when(pl.col("age_bucket_primary_filled").is_not_null())
      .then(pl.col("age_bucket_primary_filled"))
      .when(pl.col("desc_age_bucket").is_not_null())
      .then(pl.col("desc_age_bucket"))
      .otherwise(pl.lit(None))
      .alias("age_bucket_enriched"),
    pl.when(pl.col("age_bucket_primary_filled").is_not_null()).then(pl.lit("primary"))
     .when(pl.col("desc_age_bucket").is_not_null()).then(pl.lit("from_description"))
     .otherwise(pl.lit("missing"))
     .alias("age_source_enriched")
])

# 6) báo cáo
report = (
    df.select([
        pl.count().alias("total"),
        pl.col("age_bucket_primary_filled").is_not_null().sum().alias("present_primary"),
        pl.col("age_bucket_enriched").is_not_null().sum().alias("present_enriched"),
        (pl.col("age_source_enriched") == "from_description").sum().alias("filled_from_description"),
    ])
    .with_columns(
        primary_pct   = (pl.col("present_primary")   / pl.col("total") * 100).round(2),
        enriched_pct  = (pl.col("present_enriched")  / pl.col("total") * 100).round(2),
        added_by_desc = (pl.col("filled_from_description") / pl.col("total") * 100).round(2),
    )
)
print(">>> Age coverage (before/after + added by description)")
print(report)

# vài dòng mẫu điền từ mô tả
print("\n>>> Sample rows filled from description:")
print(
    df.filter(pl.col("age_source_enriched") == "from_description")
      .select(["desc_age_bucket","desc_age_confidence","description_merged"])
      .head(5)
)

# lưu lại
item_df = df
```

Điền vào cột mới age_bucket_final (ưu tiên age_bucket_enriched → cân nặng).

```python
import polars as pl

df = item_df

# 0) Chuẩn hoá mô tả (nếu chưa có _desc_norm từ trước)
if "_desc_norm" not in df.columns:
    df = df.with_columns(
        pl.col("description_merged").cast(pl.Utf8)
          .str.to_lowercase()
          .str.replace_all('"', '').str.replace_all('“|”', '')
          .str.replace_all(r"\s+", " ").str.strip_chars()
          .alias("_desc_norm")
    )

# 1) Trích cân nặng (kg)
#   - Hyphen/ndashes: "4-8kg" hoặc "4 - 8 kg"
kg_r_lo = pl.col("_desc_norm").str.extract(r'(\d+(?:[.,]\d+)?)\s*[-–—]\s*(\d+(?:[.,]\d+)?)\s*kg\b', 1)
kg_r_hi = pl.col("_desc_norm").str.extract(r'(\d+(?:[.,]\d+)?)\s*[-–—]\s*(\d+(?:[.,]\d+)?)\s*kg\b', 2)

#   - "từ ... đến ... kg"
kg_tu_lo = pl.col("_desc_norm").str.extract(r'(?:từ)\s*(\d+(?:[.,]\d+)?)\s*(?:kg)?\s*(?:đến|to|-|–|—)\s*(\d+(?:[.,]\d+)?)\s*kg\b', 1)
kg_tu_hi = pl.col("_desc_norm").str.extract(r'(?:từ)\s*(\d+(?:[.,]\d+)?)\s*(?:kg)?\s*(?:đến|to|-|–|—)\s*(\d+(?:[.,]\d+)?)\s*kg\b', 2)

#   - Cân nặng đơn: "8 kg"
kg_single = pl.col("_desc_norm").str.extract(r'\b(\d+(?:[.,]\d+)?)\s*kg\b', 1)

def to_float(expr: pl.Expr) -> pl.Expr:
    return expr.str.replace_all(",", ".").cast(pl.Float64)

df = df.with_columns([
    pl.coalesce([to_float(kg_r_lo), to_float(kg_tu_lo)]).alias("kg_lo"),
    pl.coalesce([to_float(kg_r_hi), to_float(kg_tu_hi)]).alias("kg_hi"),
    to_float(kg_single).alias("kg_single"),
])

# Nếu có khoảng thì dùng khoảng; nếu không thì dùng đơn lẻ
df = df.with_columns([
    pl.when(pl.col("kg_lo").is_not_null() & pl.col("kg_hi").is_not_null())
      .then((pl.col("kg_lo") + pl.col("kg_hi")) / 2)
      .when(pl.col("kg_lo").is_not_null() & pl.col("kg_hi").is_null())
      .then(pl.col("kg_lo"))
      .when(pl.col("kg_lo").is_null() & pl.col("kg_hi").is_null() & pl.col("kg_single").is_not_null())
      .then(pl.col("kg_single"))
      .otherwise(None)
      .alias("kg_mid")
])

# 2) Map kg_mid -> age bucket theo heuristic
def bucket_from_kg_expr(kg_mid: pl.Expr) -> pl.Expr:
    return (
        pl.when(kg_mid.is_null()).then(pl.lit(None))
         .when(kg_mid < 5.5).then(pl.lit("0-6M"))
         .when(kg_mid < 7.5).then(pl.lit("3-6M"))   # chỉ dùng nội bộ; sẽ map tiếp sang lưới 5-bucket
         .when(kg_mid < 9.5).then(pl.lit("6-9M"))
         .when(kg_mid < 11.5).then(pl.lit("9-12M"))
         .when(kg_mid < 14).then(pl.lit("1-2Y"))
         .when(kg_mid < 20).then(pl.lit("2-4Y"))
         .otherwise(pl.lit(">4Y"))
    )

df = df.with_columns(
    bucket_from_kg_expr(pl.col("kg_mid")).alias("weight_age_bucket_raw")
)

# (Tuỳ chọn) gom về lưới bucket chính 5 mức của bạn
def normalize_to_primary(bucket_col: pl.Expr) -> pl.Expr:
    return (
        pl.when(bucket_col.is_null()).then(pl.lit(None))
         .when(bucket_col == "0-6M").then(pl.lit("0-6M"))
         .when((bucket_col == "3-6M") | (bucket_col == "6-9M")).then(pl.lit("6-12M"))
         .when(bucket_col == "9-12M").then(pl.lit("6-12M"))
         .when(bucket_col == "1-2Y").then(pl.lit("1-2Y"))
         .when(bucket_col == "2-4Y").then(pl.lit("2-4Y"))
         .otherwise(pl.lit(">4Y"))
    )

df = df.with_columns(
    normalize_to_primary(pl.col("weight_age_bucket_raw")).alias("weight_age_bucket")
)

# 3) Confidence & source (range > single)
df = df.with_columns(
    pl.when(pl.col("kg_lo").is_not_null() & pl.col("kg_hi").is_not_null()).then(pl.lit(0.70))
     .when(pl.col("kg_mid").is_not_null()).then(pl.lit(0.55))
     .otherwise(pl.lit(None)).alias("weight_age_confidence"),
).with_columns(
    pl.when(pl.col("kg_lo").is_not_null() & pl.col("kg_hi").is_not_null()).then(pl.lit("weight_range"))
     .when(pl.col("kg_mid").is_not_null()).then(pl.lit("weight_single"))
     .otherwise(pl.lit(None)).alias("weight_age_source")
)

# 4) Điền bổ sung vào tuổi (sau bước enrich từ description)
df = df.with_columns([
    pl.when(pl.col("age_bucket_enriched").is_not_null())
      .then(pl.col("age_bucket_enriched"))
      .when(pl.col("weight_age_bucket").is_not_null())
      .then(pl.col("weight_age_bucket"))
      .otherwise(pl.lit(None))
      .alias("age_bucket_final"),
    pl.when(pl.col("age_bucket_enriched").is_not_null()).then(pl.lit("enriched"))
     .when(pl.col("weight_age_bucket").is_not_null()).then(pl.lit("from_weight"))
     .otherwise(pl.lit("missing"))
     .alias("age_source_final")
])

# 5) Báo cáo uplift từ cân nặng
report_w = (
    df.select([
        pl.len().alias("total"),
        pl.col("age_bucket_enriched").is_not_null().sum().alias("present_before"),
        pl.col("age_bucket_final").is_not_null().sum().alias("present_after"),
        (pl.col("age_source_final") == "from_weight").sum().alias("filled_from_weight"),
    ])
    .with_columns(
        before_pct = (pl.col("present_before") / pl.col("total") * 100).round(2),
        after_pct  = (pl.col("present_after")  / pl.col("total") * 100).round(2),
        added_pct  = (pl.col("filled_from_weight") / pl.col("total") * 100).round(2),
    )
)
print(">>> Weight-based enrichment report")
print(report_w)

print("\n>>> Sample rows filled from weight:")
print(
    df.filter(pl.col("age_source_final") == "from_weight")
      .select(["kg_lo","kg_hi","kg_mid","weight_age_bucket","weight_age_confidence","_desc_norm"])
      .head(5)
)

# Lưu lại
item_df = df
```

```python
import polars as pl
import re, json

# ---------- Helpers: normalize & parse with debug ----------
NULLISH = {"", " ", "  ", "không xác định", "khong xac dinh", "unknown", "null", "none", "nan", "na", "n/a"}

def _norm_age(s: str) -> str:
    s = s.lower().strip()
    s = (s.replace("tháng","m").replace("thang","m")
           .replace("tuổi","y").replace("tuoi","y")
           .replace("từ","tu").replace("trên","tren")
           .replace("–","-").replace("—","-").replace("−","-")
           .replace(" ", ""))
    # gom newborn
    s = (s.replace("sơsinh","nb")
           .replace("sosinh","nb")
           .replace("newborn","nb"))
    # toddler 'T' => year
    s = re.sub(r"(?<=\d)t\b", "y", s)
    return s

def _parse_age_to_months_debug(val: str | None):
    """
    Trả dict: {lo, hi, norm, reason}
    reason ∈ {nullish, adult, newborn, range, single, threshold, tokens_scan, list_scan, unparsed}
    """
    if val is None:
        return {"lo": None, "hi": None, "norm": None, "reason": "nullish"}

    raw = str(val).strip()
    if raw.lower() in NULLISH or raw == "":
        return {"lo": None, "hi": None, "norm": None, "reason": "nullish"}

    # adult product
    if raw.lower() in {"mẹ", "me"}:
        return {"lo": None, "hi": None, "norm": _norm_age(raw), "reason": "adult"}

    # list-string JSON: ["Từ 0M","Từ 3M",...]
    if raw.lstrip().startswith("["):
        try:
            items = json.loads(raw)
            lows, highs = [], []
            for it in items:
                r = _parse_age_to_months_debug(it)
                if r["lo"] is not None:
                    lows.append(r["lo"])
                if r["hi"] is not None:
                    highs.append(r["hi"])
            if lows or highs:
                lo = min(lows) if lows else None
                hi = max(highs) if highs else None
                return {"lo": lo, "hi": hi, "norm": _norm_age(raw), "reason": "list_scan"}
            else:
                return {"lo": None, "hi": None, "norm": _norm_age(raw), "reason": "list_scan"}
        except Exception:
            # sẽ rơi xuống dưới và quét token
            pass

    s = _norm_age(raw)

    # newborn
    if s == "nb":
        return {"lo": 0, "hi": 3, "norm": s, "reason": "newborn"}

    # 1) range với đơn vị ở hai đầu (cho phép trộn m/y)
    m = re.fullmatch(r"(\d+)(m|y)-(\d+)(m|y)", s)
    if m:
        a, u1, b, u2 = m.groups()
        a = int(a) * (1 if u1 == "m" else 12)
        b = int(b) * (1 if u2 == "m" else 12)
        if a > b: a, b = b, a
        return {"lo": a, "hi": b, "norm": s, "reason": "range"}

    # 2) single
    m = re.fullmatch(r"(\d+)(m|y)", s)
    if m:
        a, u = m.groups()
        a = int(a) * (1 if u == "m" else 12)
        return {"lo": a, "hi": a, "norm": s, "reason": "single"}

    # 3) threshold 'tu'/'tren'
    m = re.fullmatch(r"(tu|tren)(\d+)(m|y)", s)
    if m:
        v = int(m.group(2)) * (1 if m.group(3) == "m" else 12)
        return {"lo": v, "hi": None, "norm": s, "reason": "threshold"}

    # 4) quét token tự do trong chuỗi (fallback)
    toks = re.findall(r"(\d+)(m|y)", s)
    if toks:
        vals = [int(n) * (1 if u == "m" else 12) for (n, u) in toks]
        lo = min(vals) if vals else None
        hi = max(vals) if len(vals) > 1 else (vals[0] if vals else None)
        return {"lo": lo, "hi": hi, "norm": s, "reason": "tokens_scan"}

    # không parse được
    return {"lo": None, "hi": None, "norm": s, "reason": "unparsed"}

# ---------- Build debug table ----------
dbg = (
    item_df
    .select([
        pl.col("category_l1"),
        pl.col("age_group"),
        pl.col("age_group").map_elements(
            _parse_age_to_months_debug,
            return_dtype=pl.Struct([
                pl.Field("lo", pl.Int64),
                pl.Field("hi", pl.Int64),
                pl.Field("norm", pl.Utf8),
                pl.Field("reason", pl.Utf8),
            ])
        ).alias("p")
    ])
    .with_columns([
        pl.col("p").struct.field("lo").alias("age_lo_m"),
        pl.col("p").struct.field("hi").alias("age_hi_m"),
        pl.col("p").struct.field("norm").alias("age_norm"),
        pl.col("p").struct.field("reason").alias("parse_reason"),
    ])
    .with_columns(
        (pl.col("age_lo_m").is_null() & pl.col("age_hi_m").is_null()).alias("parse_fail")
    )
)

# ---------- 1) Tỷ lệ fail tổng quan ----------
summary = dbg.select([
    pl.len().alias("rows"),
    pl.col("parse_fail").cast(pl.Int64).sum().alias("fail_rows"),
    (pl.col("parse_fail").cast(pl.Int64).mean()*100).round(2).alias("fail_rate_%"),
    pl.col("parse_reason").filter(pl.col("parse_fail")).value_counts().alias("fail_by_reason")
])
print(summary)

# ---------- 2) TOP 20 giá trị không parse (overall) ----------
top20_fail = (
    dbg.filter(pl.col("parse_fail"))
       .group_by(["age_group","age_norm","parse_reason"])
       .agg(pl.len().alias("n"),
            pl.col("category_l1").value_counts().alias("cats"))
       .sort("n", descending=True)
       .head(20)
)
print("\nTOP 20 age_group không parse:")
print(top20_fail)

# ---------- 3) TOP 10 theo từng category_l1 ----------
top10_per_cat = (
    dbg.filter(pl.col("parse_fail"))
       .group_by(["category_l1","age_group","age_norm","parse_reason"])
       .agg(pl.len().alias("n"))
       .sort(["category_l1","n"], descending=[False, True])
       .group_by("category_l1").head(10)
)
print("\nTOP 10 không parse theo category_l1:")
print(top10_per_cat)
```

```python
import polars as pl

df = item_df  # đổi tên nếu bạn dùng df_item
FASHION = "Thời trang"

def is_missing(col: str) -> pl.Expr:
    return pl.col(col).is_null() | (
        pl.col(col).cast(pl.Utf8).str.to_lowercase().str.strip_chars() == "không xác định"
    )

# 1) Filter category Thời trang
fashion = df.filter(pl.col("category_1") == FASHION)

size_ok = ~is_missing("size")
age_ok  = ~is_missing("age_group")

# ===== A) size -> age_group =====
both = fashion.filter(size_ok & age_ok).select(["size", "age_group"])

# Đếm cặp (size, age_group)
pair = both.group_by(["size", "age_group"]).len().rename({"len":"n"})
# Tổng theo size
size_tot = both.group_by("size").len().rename({"len":"size_n"})

# Bảng ánh xạ: với mỗi size, lấy age_group đa số + tỉ lệ
map_size_to_age = (
    pair.join(size_tot, on="size")
        .with_columns((pl.col("n") / pl.col("size_n")).alias("share"))
        .group_by("size")
        .agg([
            pl.col("age_group").sort_by("share").last().alias("best_age"),
            pl.col("share").max().alias("best_share"),
            pl.col("size_n").max().alias("support"),
        ])
        .sort("best_share", descending=True)
)

# Ngưỡng chấp nhận
TAU = 0.90
MIN_SUP = 20
good_maps = map_size_to_age.filter((pl.col("best_share") >= TAU) & (pl.col("support") >= MIN_SUP))

# Ước tính số dòng có thể điền: age_group thiếu nhưng size có, và size nằm trong good_maps
can_fill_age = (
    fashion.filter(~age_ok & size_ok)
           .join(good_maps.select(["size","best_age"]), on="size", how="inner")
)
print("size -> age_group | fillable rows:", can_fill_age.height)

# (tuỳ chọn) xem vài mapping tốt
print(good_maps.head(10))

# ===== B) age_group -> size (nếu muốn điền ngược) =====
pair2 = both.group_by(["age_group","size"]).len().rename({"len":"n"})
age_tot = both.group_by("age_group").len().rename({"len":"age_n"})

map_age_to_size = (
    pair2.join(age_tot, on="age_group")
         .with_columns((pl.col("n") / pl.col("age_n")).alias("share"))
         .group_by("age_group")
         .agg([
             pl.col("size").sort_by("share").last().alias("best_size"),
             pl.col("share").max().alias("best_share"),
             pl.col("age_n").max().alias("support"),
         ])
         .sort("best_share", descending=True)
)

good_maps_rev = map_age_to_size.filter((pl.col("best_share") >= TAU) & (pl.col("support") >= MIN_SUP))

can_fill_size = (
    fashion.filter(~size_ok & age_ok)
           .join(good_maps_rev.select(["age_group","best_size"]), on="age_group", how="inner")
)
print("age_group -> size | fillable rows:", can_fill_size.height)
```
