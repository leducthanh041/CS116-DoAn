```python
import polars as pl

path = "/content/recommendation_dataset/sales_pers.item_chunk_0.parquet"
df = pl.read_parquet(path)

print(df.shape)
print(df.columns)
df.head()
```

Bước 1: Gom nhóm giá các mặt hàng có category_l1 = 'sữa' theo các nhãn "Bình dân", "Trung cấp" và "cao cấp".

```python
# Lọc đúng category_l1 = "Sữa"
milk_df = df.filter(pl.col("category_l1") == "Sữa")

# Đảm bảo cột price là số thực (float)
milk_df = milk_df.with_columns(
    pl.col("price").cast(pl.Float64)
)

# Thống kê mô tả chi tiết
stats = milk_df.select("price").describe()
print(stats)
```

```python
import polars as pl
from sklearn.cluster import KMeans
import numpy as np

# Lọc các sản phẩm có category_l1 = "Sữa"
milk_df = df.filter(pl.col("category_l1") == "Sữa")

# Đảm bảo cột price là số thực (float)
milk_df = milk_df.with_columns(pl.col("price").cast(pl.Float64))

# Gom cụm giá bằng K-Means (3 cụm)
X = np.array(milk_df["price"]).reshape(-1, 1)
kmeans = KMeans(n_clusters=3, random_state=42, n_init="auto")
labels = kmeans.fit_predict(X)

# Thêm cột "price_segment" chứa nhãn cụm
milk_df = milk_df.with_columns(pl.Series("price_segment", labels))

# Tính trung bình giá của từng cụm
cluster_mean = (
    milk_df.group_by("price_segment")
           .agg(pl.col("price").mean().alias("mean_price"))
           .sort("mean_price")
)

# Tạo mapping từ cluster → nhãn 1,2,3 (tăng dần theo giá trung bình)
mapping = {row["price_segment"]: i + 1 for i, row in enumerate(cluster_mean.iter_rows(named=True))}

# Tạo Series mới để gán nhãn 1,2,3 dựa trên mapping
price_segment_mapped = [mapping[c] for c in milk_df["price_segment"].to_list()]
milk_df = milk_df.with_columns(pl.Series("price_segment", price_segment_mapped))

# In ra kết quả gồm category_l1, price, price_segment
print("Kết quả gom cụm theo giá cho category_l1 = 'Sữa':")
milk_df.select(["category_l1", "price", "price_segment"]).head(10)
```

```python
# Đếm số lượng sản phẩm thuộc từng nhóm price_segment
segment_counts = (
    milk_df
    .group_by("price_segment")
    .agg(pl.count().alias("so_san_pham"))
    .sort("price_segment")
)

print("\nTổng số sản phẩm theo từng nhóm giá:")
print(segment_counts)
```

```python
# Thống kê min, max, mean, median của từng cụm giá
check_stats = (
    milk_df
    .group_by("price_segment")
    .agg([
        pl.count().alias("so_mat_hang"),
        pl.col("price").min().alias("min_price"),
        pl.col("price").median().alias("median_price"),
        pl.col("price").mean().alias("mean_price"),
        pl.col("price").max().alias("max_price"),
    ])
    .sort("price_segment")
)

print("\nThống kê giá theo từng cụm K-Means:")
print(check_stats)
```

Bước 2: Thống kê xem khách hàng đó mua sữa thuộc nhóm nào nhiều nhất để gán nhãn cho khách hàng đó.

```python
import polars as pl
import glob

path = "/content/recommendation_dataset/"
files = sorted(glob.glob(path + "sales_pers.purchase_history_daily_chunk_*.parquet"))


print(f"Tìm thấy {len(files)} file history")

# Đọc và gộp dữ liệu chỉ với 3 cột cần thiết
dfs = []

for f in files:
    df_chunk = pl.read_parquet(f, columns=["customer_id", "item_id", "quantity"])
    dfs.append(df_chunk)

# Gộp toàn bộ thành 1 dataframe
history_df = pl.concat(dfs)

print("Hoàn tất! Tổng số dòng:", history_df.shape[0])
history_df.head(10)
```

```python
# 1. Giữ lại chỉ các dòng liên quan đến sản phẩm sữa
#    (join theo item_id, chỉ lấy các item_id có trong milk_df)
join_df = history_df.join(
    milk_df.select(["item_id", "price_segment"]),
    on="item_id",
    how="inner"
)

print("Sau khi join, số dòng:", join_df.shape[0])
```

```python
join_df
```

```python
# 2. Tính tổng số lượng mua theo customer_id và price_segment
agg_df = (
    join_df
    .group_by(["customer_id", "price_segment"])
    .agg(pl.col("quantity").sum().alias("total_quantity"))
)
```

```python
# Bảng tổng hợp gốc
print("\nBảng tổng số lượng sữa theo từng khách hàng và phân khúc:")
print(agg_df.sort("customer_id").head(30))

# 1. Top khách hàng mua nhiều nhất
print("\n=== Top 10 customer_id có tổng quantity cao nhất ===")
top10 = (
    agg_df.group_by("customer_id")
          .agg(pl.col("total_quantity").sum().alias("tong_quantity"))
          .sort("tong_quantity", descending=True)
          .head(10)
)
top10_join = agg_df.join(top10, on="customer_id", how="inner")
print(top10_join.sort(["tong_quantity", "customer_id"], descending=[True, False]).head(30))

# 2. 20 khách hàng ngẫu nhiên
print("\n=== In ra 10 customer_id ngẫu nhiên ===")
sample20 = agg_df.sample(n=10, seed=42)
print(sample20.sort("customer_id"))

# 3. Khách hàng có đủ 3 phân khúc (1, 2, 3)
print("\n=== Khách hàng có đủ 3 phân khúc sữa khác nhau ===")
multi_seg = (
    agg_df.group_by("customer_id")
          .agg(pl.col("price_segment").n_unique().alias("so_phan_khuc"))
          .filter(pl.col("so_phan_khuc") == 3)
)
multi_seg_join = agg_df.join(multi_seg.select("customer_id"), on="customer_id", how="inner")
print(multi_seg_join.sort("customer_id").head(30))
```

```python
# 3. Với mỗi customer_id, chọn phân khúc sữa có quantity lớn nhất
#    => Đây là phân khúc "ưa thích" của khách hàng
milk_pref_df = (
    agg_df
    .sort(["customer_id", "total_quantity"], descending=[False, True])
    .group_by("customer_id")
    .agg(pl.col("price_segment").first().alias("milk_segment_preference"))
)

```

```python
print("Hoàn tất! Số khách hàng được gán nhãn:", milk_pref_df.shape[0])
milk_pref_df.head(10)
```

```python
# Thống kê số lượng khách hàng theo phân khúc sữa ưa thích
print(milk_pref_df["milk_segment_preference"].value_counts().sort("milk_segment_preference"))
```
